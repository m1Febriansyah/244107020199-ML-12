## Kegunaan Masing-Masing Library

1. **PyPREP:** Library Python untuk standardisasi dan pra-pemrosesan (*preprocessing*) sinyal EEG (rekam otak). Digunakan untuk membersihkan sinyal, mendeteksi saluran yang rusak (*bad channels*), dan menghilangkan interferensi artefak.
2. **Scipy:** Library dasar untuk komputasi ilmiah dan matematis di Python, menyediakan fungsi optimasi, pemrosesan sinyal, aljabar linier, dan analisis statistik.
3. **wandb (Weights & Biases):** Platform pelacakan eksperimen (*experiment tracking*) untuk memantau metrik pelatihan model Machine Learning, hiperparameter, dan visualisasi performa.
4. **pyECG:** Library khusus untuk analisis, ekstraksi fitur, dan pemrosesan sinyal elektrokardiogram (ECG/rekam jantung).
