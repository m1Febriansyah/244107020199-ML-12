```python
	import numpy as np
	import pandas as pd

	df = pd.read_csv('spam.csv', encoding='latin-1') # spesifiksi encoding diperlukan karena data tidak menggunakan UTF-8

	df.head()
```

```python
# Drop 3 kolom terakhir dengan fungsi iloc
df = df.drop(df.iloc[:,2:], axis=1)

# Ubah nama kolom v1 menjadi Labels, dan v2 menjadi SMS
df = df.rename(columns={'v1': 'Labels', 'v2': 'SMS'})

# Cek data
df.head()
```

```python
# Cek Jumlah Data Per Kelas
print(df['Labels'].value_counts())
print('\n')

# Cek Kelengkapan Data
print(df.info())
print('\n')

# Cek Statistik Deskriptif
print(df.describe())
```

```python
	# Data untuk label
	new_labels = {
	    'spam': 1,
	    'ham': 0
	}

	# Encode label
	df['Labels'] = df['Labels'].map(new_labels)

	# Cek data
	df.head()
```

```python
X = df['SMS'].values
y = df['Labels'].values

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer

# Split data training dan testing
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=50)

# Inisiasi CountVectorizer
bow = CountVectorizer()

# Fitting dan transform X_train dengan CountVectorizer
X_train = bow.fit_transform(X_train)

# Transform X_test
X_test = bow.transform(X_test)

# Pada scikit-learn versi baru, fungsi yang digunakan adalah get_feature_names_out()
print(len(bow.get_feature_names_out()))
print(f'Dimensi data: {X_train.shape}')
```

```python
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score

# Inisiasi MultinomialNB
mnb = MultinomialNB()

# Fit model
mnb.fit(X_train, y_train)

# Prediksi dengan data training
y_pred_train = mnb.predict(X_train)

# Evaluasi akurasi data training
acc_train = accuracy_score(y_train, y_pred_train)

# Prediksi dengan data testing
y_pred_test = mnb.predict(X_test)

# Evaluasi akurasi data testing
acc_test = accuracy_score(y_test, y_pred_test)

# Print hasil evaluasi
print(f'Hasil akurasi data train: {acc_train}')
print(f'Hasil akurasi data test: {acc_test}')
```

```python
# Load data
import pandas as pd
import numpy as np

df = pd.read_csv('properti_multi_kota.csv')
df.head()
```

```python
# Karena dataset properti adalah data tabular regresi, kita buat label target klasifikasi
# (misalnya kategori harga: Murah, Sedang, Mahal) dan gabungkan teks fitur (lokasi + sumber)
df['kategori_harga'] = pd.qcut(df['harga'], q=3, labels=['Murah', 'Sedang', 'Mahal'])

# Membuat kolom teks gabungan untuk simulasi Naive Bayes Multinomial (seperti teks SMS)
df['teks_properti'] = "Lokasi di " + df['lokasi'].astype(str) + " dari sumber " + df['sumber'].astype(str)

# Pilih kolom yang relevan
df_text = df[['teks_properti', 'kategori_harga']].dropna()
df_text.head()
```

```python
# Cek Jumlah Data Per Kelas
print(df_text['kategori_harga'].value_counts())
print('\n')

# Cek Kelengkapan Data
print(df_text.info())
```

```python
# Data untuk label mapping
new_labels = {
    'Murah': 0,
    'Sedang': 1,
    'Mahal': 2
}

# Encode label
df_text['label_encoded'] = df_text['kategori_harga'].map(new_labels)
df_text.head()
```

```python
X = df_text['teks_properti'].values
y = df_text['label_encoded'].values
```

```python
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer

# Split data training dan testing (80:20)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=50)

# Inisiasi CountVectorizer
bow = CountVectorizer()

# Fitting dan transform X_train dengan CountVectorizer
X_train = bow.fit_transform(X_train)

# Transform X_test (mencegah data leaking)
X_test = bow.transform(X_test)

print(f"Jumlah fitur kata/token: {len(bow.get_feature_names_out())}")
print(f'Dimensi data training: {X_train.shape}')
```

```python
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# Inisiasi MultinomialNB
mnb = MultinomialNB()

# Fit model
mnb.fit(X_train, y_train)

# Prediksi dengan data training & testing
y_pred_train = mnb.predict(X_train)
y_pred_test = mnb.predict(X_test)

# Evaluasi akurasi
acc_train = accuracy_score(y_train, y_pred_train)
acc_test = accuracy_score(y_test, y_pred_test)

# Print hasil evaluasi
print(f'Hasil akurasi data train: {acc_train:.4f}')
print(f'Hasil akurasi data test: {acc_test:.4f}')
print('\nClassification Report (Test Data):\n', classification_report(y_test, y_pred_test))
```
