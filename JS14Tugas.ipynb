```python
import tensorflow as tf
from tensorflow.keras import datasets, layers, models
import matplotlib.pyplot as plt

# 1. Load dan normalisasi dataset CIFAR-10
(train_images, train_labels), (test_images, test_labels) = datasets.cifar10.load_data()
train_images, test_images = train_images / 255.0, test_images / 255.0

# 2. Layer Augmentasi Data (berjalan otomatis hanya saat training)
data_augmentation = models.Sequential([
    layers.RandomFlip("horizontal"),
    layers.RandomTranslation(0.1, 0.1),
])

# 3. Bangun arsitektur CNN yang dimodifikasi
model_cifar = models.Sequential([
    layers.Input(shape=(32, 32, 3)),
    data_augmentation,

    # Blok Konvolusi 1 (32 filter)
    layers.Conv2D(32, (3, 3), padding='same', activation='relu'),
    layers.BatchNormalization(),
    layers.Conv2D(32, (3, 3), padding='same', activation='relu'),
    layers.BatchNormalization(),
    layers.MaxPooling2D((2, 2)),
    layers.Dropout(0.2),

    # Blok Konvolusi 2 (64 filter)
    layers.Conv2D(64, (3, 3), padding='same', activation='relu'),
    layers.BatchNormalization(),
    layers.Conv2D(64, (3, 3), padding='same', activation='relu'),
    layers.BatchNormalization(),
    layers.MaxPooling2D((2, 2)),
    layers.Dropout(0.3),

    # Blok Konvolusi 3 (128 filter)
    layers.Conv2D(128, (3, 3), padding='same', activation='relu'),
    layers.BatchNormalization(),
    layers.Conv2D(128, (3, 3), padding='same', activation='relu'),
    layers.BatchNormalization(),
    layers.MaxPooling2D((2, 2)),
    layers.Dropout(0.4),

    # Fully Connected Layer
    layers.Flatten(),
    layers.Dense(128, activation='relu'),
    layers.BatchNormalization(),
    layers.Dropout(0.5),
    layers.Dense(10)
])

# 4. Kompilasi model
model_cifar.compile(
    optimizer='adam',
    loss=tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True),
    metrics=['accuracy']
)

# 5. Latih model (25 epoch cukup untuk menembus > 80%)
history = model_cifar.fit(
    train_images, train_labels,
    epochs=25,
    batch_size=64,
    validation_data=(test_images, test_labels)
)

# 6. Evaluasi hasil akhir
test_loss, test_acc = model_cifar.evaluate(test_images, test_labels, verbose=2)
print(f"\nAkurasi Testing CIFAR-10: {test_acc * 100:.2f}%")
```

```python
import tensorflow as tf
from tensorflow.keras import datasets, layers, models

# 1. Load dataset MNIST
(X_train, y_train), (X_test, y_test) = datasets.mnist.load_data()

# 2. Reshape untuk menambahkan dimensi kanal warna (grayscale = 1) & normalisasi (0-1)
X_train = X_train.reshape((X_train.shape[0], 28, 28, 1)).astype('float32') / 255.0
X_test = X_test.reshape((X_test.shape[0], 28, 28, 1)).astype('float32') / 255.0

# 3. Bangun arsitektur CNN MNIST
model_mnist = models.Sequential([
    layers.Input(shape=(28, 28, 1)),

    # Layer Konvolusi & Pooling 1
    layers.Conv2D(32, kernel_size=(3, 3), activation='relu'),
    layers.MaxPooling2D(pool_size=(2, 2)),

    # Layer Konvolusi & Pooling 2
    layers.Conv2D(64, kernel_size=(3, 3), activation='relu'),
    layers.MaxPooling2D(pool_size=(2, 2)),

    # Flattening & Fully Connected Layer
    layers.Flatten(),
    layers.Dropout(0.5),
    layers.Dense(128, activation='relu'),
    layers.Dense(10, activation='softmax')
])

model_mnist.summary()

# 4. Kompilasi model
# Menggunakan sparse_categorical_crossentropy karena label y masih berupa integer (0-9)
model_mnist.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# 5. Latih model
history_mnist = model_mnist.fit(
    X_train, y_train,
    epochs=10,
    batch_size=64,
    validation_split=0.1
)

# 6. Evaluasi pada data uji
test_loss, test_acc = model_mnist.evaluate(X_test, y_test, verbose=2)
print(f"\nAkurasi Testing CNN MNIST: {test_acc * 100:.2f}%")
```
