```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, StandardScaler

df = pd.read_csv('Titanic-Dataset-fixed.csv')
df.head()
```

```python
df = df[['Survived', 'Pclass', 'Age', 'Sex', 'Cabin']]
df.head()
```

```python
le = LabelEncoder() # membuat objek dari LabelEncoder
df['Sex'] = le.fit_transform(df['Sex']) # proses encoding
df['Cabin'] = le.fit_transform(df['Cabin']) # proses encoding
```

```python
df.head()
```

```python
std = StandardScaler()
df['Age'] = std.fit_transform(df[['Age']])
```

```python
df.head()
```

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, StandardScaler
df = pd.read_csv('properti_multi_kota.csv')

# Imputasi cepat jika memakai file baru (agar tidak ada NaN)
for col in ['luas_bangunan', 'jumlah_kamar', 'jumlah_kamar_mandi', 'usia_bangunan']:
    df[col] = df[col].fillna(df[col].median())

df.head()
```

```python
df = df[['lokasi', 'luas_m2', 'luas_bangunan', 'jumlah_kamar', 'jumlah_kamar_mandi', 'harga']]
df.head()
```

```python
le = LabelEncoder() # membuat objek dari LabelEncoder
df['lokasi'] = le.fit_transform(df['lokasi']) # proses encoding
```

```python
df.head()
```

```python
std = StandardScaler()
fitur_numerik = ['luas_m2', 'luas_bangunan', 'jumlah_kamar', 'jumlah_kamar_mandi']
df[fitur_numerik] = std.fit_transform(df[fitur_numerik])
```
