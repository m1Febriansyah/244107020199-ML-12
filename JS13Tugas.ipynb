```python
import tensorflow as tf
from tensorflow.keras.datasets import mnist
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten
from tensorflow.keras.utils import to_categorical

# 1. Load dataset MNIST
(X_train, y_train), (X_test, y_test) = mnist.load_data()

# Normalisasi data (0-255 → 0-1)
X_train = X_train / 255.0
X_test = X_test / 255.0

# One-hot encoding label
y_train = to_categorical(y_train, 10)
y_test = to_categorical(y_test, 10)

# 2. Bangun model JST
model = Sequential([
    # Ubah gambar 28x28 menjadi vektor (1D array berisi 784 elemen)
    Flatten(input_shape=(28, 28)),

    # Hidden layer 1 dengan 128 neuron dan fungsi aktivasi ReLU
    Dense(128, activation='relu'),

    # Hidden layer 2 dengan 64 neuron dan fungsi aktivasi ReLU
    Dense(64, activation='relu'),

    # Output layer (10 kelas untuk digit 0-9) dengan fungsi aktivasi Softmax
    Dense(10, activation='softmax')
])

# 3. Kompilasi model
# Menggunakan categorical_crossentropy karena label sudah di-one-hot encoding
model.compile(optimizer='adam',
              loss='categorical_crossentropy',
              metrics=['accuracy'])

# 4. Latih model
# Menggunakan validation_split untuk memantau performa pada data validasi (20% dari data latih)
print("Memulai pelatihan model...")
history = model.fit(X_train, y_train, epochs=10, batch_size=32, validation_split=0.2)

# 5. Evaluasi model
print("\nEvaluasi pada data pengujian:")
loss, acc = model.evaluate(X_test, y_test)
print(f"Akurasi pada data uji: {acc:.4f}")
```
