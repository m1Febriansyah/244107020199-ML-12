```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder, StandardScaler

df = pd.read_csv('wbc.csv')

kolom_tidak_terpakai = ['id', 'Unnamed: 32']
df_berguna = df.drop(columns=kolom_tidak_terpakai)

df.head()
```

```python
encoder = LabelEncoder()
df_berguna['diagnosis'] = encoder.fit_transform(df_berguna['diagnosis'])

df.head()
```

```python
kolom_numerik = df_berguna.drop('diagnosis', axis=1).columns

scaler = StandardScaler()
df_berguna[kolom_numerik] = scaler.fit_transform(df_berguna[kolom_numerik])

print(df_berguna.head())
```
