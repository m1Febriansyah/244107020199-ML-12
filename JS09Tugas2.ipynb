```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# Membaca dataset (Gunakan encoding latin-1 untuk spam.csv)
df = pd.read_csv('spam.csv', encoding='latin-1')

# Membersihkan kolom yang tidak berguna dan mengubah nama kolom
df = df.drop(['Unnamed: 2', 'Unnamed: 3', 'Unnamed: 4'], axis=1)
df.columns = ['label', 'text']

# Mengubah label menjadi numerik: ham = 0, spam = 1
df['label'] = df['label'].map({'ham': 0, 'spam': 1})

# Membagi data menjadi data latih (80%) dan data uji (20%)
X_train, X_test, y_train, y_test = train_test_split(df['text'], df['label'], test_size=0.2, random_state=42)
```

```python
# Inisialisasi CountVectorizer dengan stop_words
cv = CountVectorizer(stop_words='english')

# Transformasi data teks menjadi matriks frekuensi
X_train_cv = cv.fit_transform(X_train)
X_test_cv = cv.transform(X_test)

# Inisialisasi dan melatih model Multinomial Naive Bayes
model_cv = MultinomialNB()
model_cv.fit(X_train_cv, y_train)

# Prediksi dan Evaluasi
y_pred_cv = model_cv.predict(X_test_cv)

print("=== EVALUASI MODEL COUNTVECTORIZER ===")
print(f"Akurasi: {accuracy_score(y_test, y_pred_cv):.4f}")
print("Laporan Klasifikasi:\n", classification_report(y_test, y_pred_cv, target_names=['Ham', 'Spam']))
```
