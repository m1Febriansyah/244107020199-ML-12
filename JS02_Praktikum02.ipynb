```python
import pandas as pd
import numpy as np
df = pd.read_csv('Titanic-Dataset.csv')

# Tampilkan 5 data teratas
df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
# Age - mean
df['Age'] = df['Age'].fillna(value=df['Age'].mean())

# Cabin - "DECK"
df['Cabin'] = df['Cabin'].fillna(value="DECK")

# Embarked - modus (ditambah ()[0] untuk mengambil nilai teksnya)
df['Embarked'] = df['Embarked'].fillna(value=df['Embarked'].mode()[0])
```

```python
df.info()
```

```python
df.to_csv('Titanic-Dataset-Fixed.csv', index=False)
```

```python
import pandas as pd
import numpy as np
df = pd.read_csv('properti_multi_kota.csv')

# Tampilkan 5 data teratas
df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
# 1. luas_bangunan - menggunakan median (menghindari bias outlier ekstrem)
df['luas_bangunan'] = df['luas_bangunan'].fillna(df['luas_bangunan'].median())

# 2. jumlah_kamar - menggunakan median / modus
df['jumlah_kamar'] = df['jumlah_kamar'].fillna(df['jumlah_kamar'].median())

# 3. jumlah_kamar_mandi - menggunakan median / modus
df['jumlah_kamar_mandi'] = df['jumlah_kamar_mandi'].fillna(df['jumlah_kamar_mandi'].median())

# 4. usia_bangunan - menggunakan median (atau bisa diisi 0 jika diasumsikan bangunan baru)
df['usia_bangunan'] = df['usia_bangunan'].fillna(df['usia_bangunan'].median())
```
