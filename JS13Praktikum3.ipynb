```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
import tensorflow as tf

# Contoh dataset (buat dummy data)
data = pd.DataFrame({
    'luas': [50, 60, 70, 80, 90],
    'harga': [500, 600, 700, 800, 900]
})

X = data[['luas']]
y = data[['harga']]

# Normalisasi
scaler = StandardScaler()
X = scaler.fit_transform(X)
y = scaler.fit_transform(y)

# Split data
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

# Model
model = tf.keras.Sequential([
    tf.keras.layers.Dense(10, activation='relu', input_shape=(1,)),
    tf.keras.layers.Dense(1)
])

model.compile(optimizer='adam', loss='mse')
model.fit(X_train, y_train, epochs=100)

# Evaluasi
print("Prediksi:", model.predict(X_test))
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
import tensorflow as tf

# Contoh dataset (buat dummy data)
data = pd.DataFrame({
    'luas': [50, 60, 70, 80, 90],
    'harga': [500, 600, 700, 800, 900]
})

X = data[['luas']]
y = data[['harga']]

# Gunakan dua scaler berbeda untuk fitur (X) dan target (y)
scaler_X = StandardScaler()
scaler_y = StandardScaler()

X_scaled = scaler_X.fit_transform(X)
y_scaled = scaler_y.fit_transform(y)

# Split data (test_size 0.2 dari 5 data berarti 1 data test, 4 data train)
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y_scaled, test_size=0.2, random_state=42)

# Model Regresi
model = tf.keras.Sequential([
    tf.keras.layers.Dense(10, activation='relu', input_shape=(1,)),
    tf.keras.layers.Dense(1) # Tidak memakai aktivasi tambahan (Linear) untuk regresi
])

# MSE cocok untuk regresi
model.compile(optimizer='adam', loss='mse')
model.fit(X_train, y_train, epochs=100, verbose=0) # verbose=0 agar terminal tidak penuh

# Evaluasi dan Prediksi
prediksi_scaled = model.predict(X_test)

# Kembalikan nilai prediksi ke skala harga aslinya
prediksi_asli = scaler_y.inverse_transform(prediksi_scaled)
nilai_asli_test = scaler_y.inverse_transform(y_test)

print("\n--- Hasil Uji ---")
# Menampilkan X_test ke bentuk semula (luas awal)
print(f"Luas (X_test): {scaler_X.inverse_transform(X_test)[0][0]}")
print(f"Harga Asli: {nilai_asli_test[0][0]}")
print(f"Harga Prediksi: {prediksi_asli[0][0]:.2f}")
```
